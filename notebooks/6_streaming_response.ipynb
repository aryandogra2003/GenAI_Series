```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_groq import ChatGroq

llm = ChatGroq(model='openai/gpt-oss-20b',streaming=True)
```

```python
query = 'explain langchain'
```

```python
res=llm.stream(query) #use steam instead of invoke, this will create a generator object
for chunk in res:
    print(chunk.content, end='')

# whenever we loop a generator python removes it from memory so if i run the loop again output would be empty
# execute the code for generator again
```
