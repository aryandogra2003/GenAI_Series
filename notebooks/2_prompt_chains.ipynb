```python
from langchain_groq import ChatGroq
llm = ChatGroq(model='openai/gpt-oss-20b')
```

```python
prompt=[
    {"role":"system", "content":"You are a Javascript developer"},
    {"role":"user", "content":"What is the difference between maps and filters?"},
]

res=llm.invoke(prompt)
print(res.content)
```

### Prompt Templates - Dynamic Prompts

```python
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
llm = ChatGroq(model='openai/gpt-oss-20b')
out = StrOutputParser()
```

```python
prompt =ChatPromptTemplate.from_messages([
    {"role":"system", "content":"You are a translator, and translate input in {language}"},
    {"role":"user","content":"{query}"},
])
```

```python
final_prompt = prompt.format_messages(language="Hindi",
                                      query= "I love programming")

res = llm.invoke(final_prompt)
print(res.content)
```

```python
## Both LLM and Prompt are runnable objects, so we can also do the following:
res=llm.invoke(prompt.invoke({"language":"Hindi","query":"I love programming"}))
print(res.content)
```

## Chains- Runnables

```python
chains = prompt | llm | out  # pipe operator to create a chain
chains  # chains is also a runnable object, so we can invoke it directly
```

```python
res=chains.invoke({"language":"HingLish","query":"I love langchain"})
print(res) # no need to access content attribute, as we have used StrOutputParser
```

```python
def response_upper(res:str):
    return res.upper()
```

```python
chains = prompt | llm | out | response_upper # you can API calls functions in the chain as well
res=chains.invoke({"language":"HingLish","query":"I love langchain"})
print(res) 
```
