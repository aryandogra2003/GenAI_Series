```python
from langchain_groq import ChatGroq
llm=ChatGroq(model="openai/gpt-oss-20b")
```

### Problem

```python
query='Hello, I am Aryan Dogra, my email is aryandogra2508@gmail.com and i am 23 years old'
res=llm.invoke(f"Please give me the name , email and age from {query}")

print(res.content)
# we cant select only name/email or age from this reponse, it is of type string
```

### Solution: Structured Output

```python
from pydantic import BaseModel, Field

class ResponseStructure (BaseModel):
    name:str= Field(description="complete name of user")
    email:str = Field(description="Email of user")
    age:int = Field(description="Age of user")

structuredLLM = llm.with_structured_output(ResponseStructure)
```

```python
res = structuredLLM.invoke(query)
# print(res.name)
# print(res.email)
# print(res.age)
res.model_dump()
```

```python
from typing import List

class Movies(BaseModel):
    title:str = Field(description='Movie Title')
    year:int = Field(description='release date of movie')

class AllMovies (BaseModel):
    movies: List[Movies]

movie_llm = llm.with_structured_output(AllMovies, method="json_mode")
query2 = (
    "Return a JSON object matching the requested schema. "
    "Include several horror movies, each with a title and release year."
)

res = movie_llm.invoke(query2)
print(res)
# print(res.model_dump())
```
