```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_numbers(a:int,b:int):
    """
        It will return sum of two numbers
        Args:
            a: number one
            b: number two

    """
    return a+b

@tool
def multiply_numbers(a:int,b:int):
    """
        It will return product of two numbers
        Args:
            a: number one
            b: number two

    """
    return a*b
```

```python
add_numbers.invoke({"a":12,"b":3})
```

```python
from langchain_groq import ChatGroq
from langchain.agents import create_agent

llm = ChatGroq(model='openai/gpt-oss-20b',streaming=True)
```

```python
agent = create_agent(
    model=llm,
    tools=[add_numbers, multiply_numbers],
    system_prompt='Ypu are a math teacher, and always use tool for calculation'
)
```

```python
res = agent.invoke({"messages":[{"role":"user", "content":"what is 2*3"}]})
res
```
